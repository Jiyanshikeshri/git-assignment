# Assignment 3: Data Cleaning

This notebook contains how to detect missing values and clean a dataset
using Pandas by replacing missing entries with appropriate values.

```python
# Importing the required libraries
import pandas as pd
import numpy as np
```

```python
# Creating a dataset containing missing values
employee_data = {
    "Name": ["Rahul", "Priya", "Anuj"],
    "Age": [25, np.nan, 29],
    "Salary": [30000, 40000, np.nan]
}

dataframe = pd.DataFrame(employee_data)

print("Original DataFrame:")
dataframe
```

```python
# Detecting missing values
print("Missing Values:")
dataframe.isnull()
```

```python
# Replacing missing values in the Age column with the mean value of age
average_age = dataframe["Age"].mean()

dataframe["Age"] = dataframe["Age"].fillna(average_age)

print("DataFrame after replacing missing Age:")
dataframe
```

```python
# Replacing missing salary values with 0
dataframe["Salary"] = dataframe["Salary"].fillna(0)

print("DataFrame:")
dataframe
```
