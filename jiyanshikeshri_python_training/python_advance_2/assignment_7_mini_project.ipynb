This notebook analyzes student performance using Pandas, Matplotlib, and Seaborn.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Create a DataFrame containing student information

student_data = {
    "Name": ["Rahul", "Priya", "Siri", "Anuj"],
    "Marks": [70, 80, 90, 60],
    "Hours Studied": [2, 3, 5, 1]
}

dataframe = pd.DataFrame(student_data)

dataframe
```

```python
#Adding performance column based on marks

dataframe["Performance"] = dataframe["Marks"].apply(
    lambda marks: "Pass" if marks > 65 else "Fail"
)

dataframe
```

```python
#Creating line chart between hours studied and marks

# Sorting the dataset by Hours Studied so that the line chart read chronologically
dataframe_sorted = dataframe.sort_values(by='Hours Studied')

plt.figure(figsize=(6, 4))

plt.plot(
    dataframe_sorted["Hours Studied"],
    dataframe_sorted["Marks"],
    marker="o"
)

plt.title("Hours Studied vs Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")

plt.show()
```

```python
#Creating scatter plot between hours studied and marks

plt.figure(figsize=(6, 4))

plt.scatter(
    dataframe["Hours Studied"],
    dataframe["Marks"]
)

plt.title("Study Hours vs Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")

plt.show()
```

```python
#Creating bar chart between performance and marks

plt.figure(figsize=(6, 4))

sns.barplot(
    data=dataframe,
    x="Performance",
    y="Marks"
)

plt.title("Performance vs Marks")

plt.show()
```
