# Assignment 2: Pandas DataFrame Creation

This notebook contains how to create a Pandas DataFrame, filter data, and add a new calculated column.

```python
import pandas as pd
```

```python
# Creating a dictionary containing employee information
employee_data = {
    "Name": ["Rahul", "Priya", "Amit", "Anuj"],
    "Age": [25, 30, 28, 35],
    "Department": ["HR", "IT", "Finance", "IT"],
    "Salary": [30000, 50000, 45000, 60000]
}

# Converting the dictionary into a Pandas DataFrame
dataframe = pd.DataFrame(employee_data)

print("Employee DataFrame:")
dataframe
```

```python
# Displaying the first two rows
print("First Two Rows:")
dataframe.head(2)
```

```python
# Showing summary statistics
print("Summary Statistics:")
dataframe.describe()
```

```python
# Showing employees who belong to the IT department
it_employees = dataframe[dataframe["Department"] == "IT"]

print("IT Employees:")
it_employees
```

```python
# Adding new column: Bonus = Salary * 0.10
dataframe["Bonus"] = dataframe["Salary"] * 0.10

print("Updated DataFrame with Bonus:")
dataframe
```
