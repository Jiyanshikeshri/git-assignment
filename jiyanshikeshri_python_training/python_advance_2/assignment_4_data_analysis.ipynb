# Assignment 4: Data Analysis Using GroupBy

This notebook contains how to analyze employee data using the
groupby() function in Pandas to calculate average salary, maximum salary,
and employee count for each department.

```python
import pandas as pd
```

```python
# Creating the employee dataset
employee_data = {
    "Name": ["Rahul", "Priya", "Amit", "Anuj"],
    "Age": [25, 30, 28, 35],
    "Department": ["HR", "IT", "Finance", "IT"],
    "Salary": [30000, 50000, 45000, 60000]
}

dataframe = pd.DataFrame(employee_data)

print("Employee DataFrame:")
dataframe
```

```python
# Finding the average salary by department
average_salary = dataframe.groupby("Department")["Salary"].mean()

print("Average Salary by Department:")
average_salary
```

```python
#Finding the max salary by department
max_salary = dataframe.groupby("Department")["Salary"].max()

print("Max Salary by Department:")
max_salary
```

```python
#Counting the number of employees in each department
employee_count = dataframe.groupby("Department")["Name"].count()

print("Employee Count in each Department:")
employee_count
```
