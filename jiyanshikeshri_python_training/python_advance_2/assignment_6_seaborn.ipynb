Assignment 6: Seaborn Visualizations

This notebook demonstrates different visualizations using the Seaborn library.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#Create a DataFrame containing employee details

employee_data = {
    "Name": ["Rahul", "Priya", "Amit", "Anuj"],
    "Age": [25, 30, 28, 35],
    "Department": ["HR", "IT", "Finance", "IT"],
    "Salary": [30000, 50000, 45000, 60000]
}

dataframe = pd.DataFrame(employee_data)

dataframe
```

```python
#Creating a barplot department vs salary

plt.figure(figsize=(6, 4))

sns.barplot(
    data=dataframe,
    x="Department",
    y="Salary"
)

plt.title("Department vs Salary")
plt.show()
```

```python
#Creating a boxplot to show salary distribution

plt.figure(figsize=(6, 4))

sns.boxplot(
    data=dataframe,
    y="Salary"
)

plt.title("Salary Distribution")
plt.show()
```

```python
#Creating a heatmap using correlation between Age & Salary

correlation_matrix = dataframe[["Age", "Salary"]].corr()

plt.figure(figsize=(5, 4))
# annot=True, puts the correlation numbers inside the colored boxes
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="Blues"
)

plt.title("Correlation Between Age and Salary")
plt.show()
```
